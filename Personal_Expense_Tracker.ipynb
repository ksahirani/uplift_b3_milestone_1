# Personal Expense Tracker
**Milestone 1 — Python Foundations Build**

A command-line program that lets a user log expenses (description, amount, category),
then reports a total, a per-category breakdown, the highest/lowest expense, and a
budget status.

**Core requirements covered:** variables & data types, conditionals, loops, 2+ single-purpose
functions, and (bonus) input validation with exception handling.

```python

BUDGET_LIMIT = 50000.00  # monthly budget, in currency units

    
    # Repeatedly prompts the user for an expense amount until a valid,
    # non-negative number is entered. Demonstrates a validation loop +
    # exception handling.
def get_valid_amount():

    while True:
        raw_input_value = input("  Amount: ")                                            
        try:
            amount = float(raw_input_value)
            if amount <= 0:
                print("  Amount must be greater than 0. Try again.")
                continue
            return amount
        except ValueError:
            # Catches non-numeric input, e.g. "twenty" or "12,50"
            print("  That's not a valid number. Please enter digits only (e.g. 12.50).")


    # Adds a single expense record (a dict) to the expenses list.
    # Single responsibility: build and store one record.
def add_expense(expenses, description, amount, category):
    expense_record = {
        "description": description,
        "amount": amount,
        "category": category,
    }
    expenses.append(expense_record)
    return expenses

# Sums the 'amount' field across all expense records.
def calculate_total(expenses):
    total = 0.0
    for expense in expenses:
        total += expense["amount"]
    return total

# Builds a dict of {category: total_spent_in_category}.
def summarize_by_category(expenses):
    category_totals = {}
    for expense in expenses:
        category = expense["category"]
        amount = expense["amount"]
        if category in category_totals:
            category_totals[category] += amount
        else:
            category_totals[category] = amount
    return category_totals

   # Returns a tuple of (highest_expense, lowest_expense) as dicts.
    # Demonstrates a function returning multiple values (Advanced Track #4).
    # Returns (None, None) if there are no expenses.
def get_highest_and_lowest(expenses):
    if not expenses:
        return None, None
    highest = expenses[0]
    lowest = expenses[0]
    for expense in expenses:
        if expense["amount"] > highest["amount"]:
            highest = expense
        if expense["amount"] < lowest["amount"]:
            lowest = expense
    return highest, lowest

# Uses conditional logic to classify spending against the budget.
# Returns a human-readable status string.
def get_budget_status(total, budget_limit):
    if total > budget_limit:
        over_by = total - budget_limit
        return f"Over budget by {over_by:.2f}!"
    elif total == budget_limit:
        return "Exactly at your budget limit."
    elif total >= budget_limit * 0.8:
        remaining = budget_limit - total
        return f"Getting close to your budget. {remaining:.2f} remaining."
    else:
        remaining = budget_limit - total
        return f"Within budget. {remaining:.2f} remaining."

 # Prints the final report: total, category breakdown, highest/lowest,
 # and budget status. Pulls together the other functions' results.
def print_summary(expenses, budget_limit):
   
    total = calculate_total(expenses)
    category_totals = summarize_by_category(expenses)
    highest, lowest = get_highest_and_lowest(expenses)
    status = get_budget_status(total, budget_limit)

    print("\n----- Expense Summary -----")
    print(f"Total expenses logged: {len(expenses)}")
    print(f"Total spent: {total:.2f}")

    print("\nBy category:")
    for category, amount in category_totals.items():
        print(f"  {category}: {amount:.2f}")

    if highest is not None:
        print(f"\nHighest expense: {highest['description']} ({highest['amount']:.2f})")
        print(f"Lowest expense:  {lowest['description']} ({lowest['amount']:.2f})")

    print(f"\nBudget status: {status}")
    print("----------------------------\n")

    # Main program loop: collects expenses from the user until they type
    # 'done', then prints the summary.
def run_tracker():
    expenses = []
    is_tracking = True

    print("Personal Expense Tracker")
    print("Enter your expenses one at a time. Type 'done' when finished.\n")

    while is_tracking:
        description = input("Expense description (or 'done' to finish): ")

        if description.lower() == "done":
            is_tracking = False
            continue

        if description.strip() == "":
            print("  Description can't be empty. Try again.\n")
            continue

        amount = get_valid_amount()
        category = input("  Category (e.g. Food, Transport, Bills): ").strip()
        if category == "":
            category = "Uncategorized"

        add_expense(expenses, description, amount, category)
        print(f"  Added: {description} - {amount:.2f} ({category})\n")

    if len(expenses) == 0:
        print("No expenses were logged. Goodbye!")
        return

    print_summary(expenses, BUDGET_LIMIT)
```

## Run the tracker
Run the cell below, then follow the prompts. Type `done` when you're finished entering expenses.

```python
run_tracker()
```
